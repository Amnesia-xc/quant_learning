## Example

```python
import numpy as np
a = np.arange(12)
m = a.reshape(3,4)
print(m)
print(a.reshape(-1,6))
print(a.reshape(2,2,3))
```

```python
print(m.ravel())
print(m.flatten())

x = np.zeros((3,1,4))
print(x)
print(x.shape)
print(x.squeeze())
print(x.squeeze().shape)
print(np.expand_dims(np.zeros(3),0))
print(np.expand_dims(np.zeros(3),0).shape)
```

```python
x = np.array([[1,2],[3,4]])
y = np.array([[5,6],[7,8]])
print(np.concatenate([x,y],axis=0))
print(np.concatenate([x,y],axis=1))
print(np.stack([x,y],axis=0))
print(np.stack([x,y],axis=1))
print(np.stack([x,y],axis=2))
print(np.vstack([x,y]))
print(np.hstack([x,y]))
```

```python
m = np.arange(6).reshape(2,3)
print(m)
print(m.T)
print(m.transpose(1,0))
print(np.swapaxes(np.zeros((2,3,4)),0,2))
```

```python
m = np.arange(1,13).reshape(3,4)
print(m)


print(m.sum(axis=0))
print(m.sum(axis=1))
```

```python
t = np.ones((2,3,4))
print(t.sum(axis=1))
print(t.mean(axis=(0,2)))
print(t.sum(axis=(0,2)))
```

```python
m = np.arange(1,13).reshape(3,4)

print(m.sum(axis=0).shape)
print(m.sum(axis=0,keepdims=True).shape)
print(m.sum(axis=0,keepdims=True))
```

```python
m = np.arange(1,13).reshape(3,4)
print(m.sum(axis=0,keepdims=True))
print(m.mean(axis=1))
print(m.argmax(axis=1))
```

```python
m = np.arange(1,13).reshape(3,4)
print(m + 100)

print(m + np.array([10,20,30,40]))

col = np.array([1,2,3]).reshape(3,1)
print(m + col)
print(np.arange(3).reshape(3,1) + np.arange(4).reshape(1,4))
```

```python
rng = np.random.default_rng(0)
d = rng.normal(0,1,(5,4))

col_centered = d - d.mean(axis=0,keepdims=True)
print(col_centered.mean(axis=0).round(10))

row_centered = d - d.mean(axis=1,keepdims=True)
print(row_centered.mean(axis=1).round(10))
```

```python
c1 = np.arange(3).reshape(3,1)
c2 = np.arange(0,40,10).reshape(1,4)
print(c1 + c2)
```

```python
rng = np.random.default_rng(0)
d = rng.normal(0,1,(5,4))

d_col_centered = d - d.mean(axis=0, keepdims=True)
print(d_col_centered.mean(axis=0).round(10))

d_row_centered = d - d.mean(axis=1, keepdims=True)
print(d_row_centered.mean(axis=1).round(10))
```
