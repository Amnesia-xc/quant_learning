```python
import numpy as np
rng = np.random.default_rng(21)
M = rng.standard_normal((5,3))

M_col_centered = (M - M.mean(axis=0,keepdims=True))/M.std(axis=0,keepdims=True)
print(M_col_centered)
print(M_col_centered.argmax(axis=1))
print(M_col_centered[np.argsort(M_col_centered.sum(axis=1)),:])
```
