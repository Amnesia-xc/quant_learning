```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. 验证版本
print("numpy:", np.__version__)
print("pandas:", pd.__version__)

# 2. 造一个小数据表
df = pd.DataFrame({
    "x": np.arange(10),
    "y": np.random.default_rng(42).random(10)
})
print(df.head())

# 3. 画一张图（验证 matplotlib）
plt.plot(df["x"], df["y"], marker="o")
plt.title("env test ok")
plt.show()
```
