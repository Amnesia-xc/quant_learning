## Example

```python
import numpy as np
a = np.arange(10)

print(a[0], a[-1])
print(a[1:4])
print(a[:5])
print(a[::2])
print(a[::-1])
```

```python
rng = np.random.default_rng(0)
m = rng.integers(0,10,(5,6))
print(m)

print(m[0,1])
print(m[0][1])
print(m[0,:])
print(m[:,1])
print(m[:2,:3])
print(m[1:4, ::2])
```

```python
m = np.arange(12).reshape(3,4)
print(m)
s = m[0,:]
s[0] = 99
print(s)
print(m)
print(np.shares_memory(s,m))

m = np.arange(12).reshape(3,4)
s2 = m[0,:].copy()
s2[0] = 99
print(m)
print(np.shares_memory(s2,m))
```

```python
rng = np.random.default_rng(1)
x = rng.normal(0,1,10)
print(x)
print(x > 0)
print(x[x > 0])
print((x > 0).sum())
print((x > 0).mean())
print(x[(x > 0) & (x < 1)])
print(x[~(x > 0)])

print(np.where(x > 0, x, 0))
print(np.where(x > 0))
print(np.where(x > 0)[0])
print(np.where(x > 0)[0].size)
print(np.where(x > 0)[0].dtype)
```

```python
a = np.arange(10,60,10)
print(a[[4,0,2]])
print(a[2:4])

m = np.arange(20).reshape(4,5)
print(m[[0,2],:])
print(m[:,[0,4]])

a[[0,2,4]] = 0
print(a)
```

```python
rng = np.random.default_rng(0)
b = rng.integers(0,10,(5,6))
print(b[2,:])
print(b[:,3])
print(b[:2,:3])
print(b[4,5])
```

```python

s = b[:, 0]
s[0] = 99
print(b)
print(np.shares_memory(s,b))

s = b[:, 0].copy()
s[0] = 1001
print(b)
print(np.shares_memory(s,b))
```

```python
rng = np.random.default_rng(1)
d = rng.normal(0,1,100)
print((d > 0).mean())
print(np.where(d >= 0, d, 0))
print(((d > 1) | (d < -1)).sum())
```
