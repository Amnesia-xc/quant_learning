## Example

```python
import numpy as np
rng = np.random.default_rng(0)
a = rng.normal(0,1,6)

print(np.exp(np.arange(5)))
print(np.log(np.arange(1,6)))
print(np.sqrt(np.arange(5)))
print(np.abs(a), np.sign(a))
print(np.floor(a), np.ceil(a), np.round(a,2))

print(np.maximum(np.array([1,5,3]), np.array([2,4,6])))
```

```python
m = rng.normal(0,1,(6,4))

print(m.sum(), m.prod())
print(m.mean(), np.median(m))
print(m.sum(), np.sum(m))
print(m.var(), m.std())
print(m.min(), m.max())
print(m.argmin(), m.argmax())
print(np.percentile(m, [25,50,75]))
print(m.any(), m.all())
```

```python
x = rng.normal(0,1,100)
print(x.std())
print(x.std(ddof=1))
```

```python
print(m.mean(axis=1))
print(m.std(axis=1, ddof=1))
print(m.argmax(axis=0))
print(m.sum(axis=0,keepdims=True).shape)
```

```python
print(np.arange(1,6).cumsum())
print(np.arange(1,6).cumprod())
print(np.maximum.accumulate(np.arange(1,6)))
```

```python
r = rng.normal(0,0.02,10)

print(np.where(r > 0,1,0))
print(np.where(r > 0,r,0))
print(np.where(r > 0))

labels = np.select([r > 0.02, r < -0.02], ["大涨", "大跌"], default="震荡")
print(list(zip(np.round(r,4), labels)))

print(np.clip(rng.normal(0,1,10),-2,2))
```

```python
scores = np.array([72,95,88,60,88])

print(np.sort(scores))
print(np.argsort(scores))
print(np.argsort(scores)[::-1])
print(np.sort(scores)[::-1])

tmp = scores.copy()
tmp.sort()
print(tmp)

m = rng.integers(0,100,(5,3))
order = np.argsort(m[:,0])
print(order)
print(m[order])
```
