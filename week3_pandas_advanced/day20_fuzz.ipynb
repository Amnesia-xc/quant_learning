```python
import numpy as np
import pandas as pd

cities = pd.Series(['北京','北京市','北京 ','BEIJING','北 京','上海','上海市','上诲',
                    '广州','广州市','广洲','深圳','深圳市','深川'])
```

```python
cities1 = cities.str.lower().str.replace(' ','')

def map_city(v):
    m, s = process.extractOne(v,standard)
    return m if s >= 90 else None

cities2 = cities1.map(map_city)

manual = {'beijing':'北京','上诲':'上海','广洲':'广州','深川':'深圳'}
cities3 = cities2.fillna(cities1.map(manual))
print(cities3)
```

```python
def fuzzy_standardize(series, standards, threshold):
    series = series.str.lower().str.replace(' ','')
    
    def map_city(v):
        m, score = process.extractOne(v, standards)
        return m if score >= threshold else None
    
    cleaned = series.map(map_city)
    
    check = series[cleaned.isna()]
    return [cleaned, check]

fuzzy_standardize(cities, standard, 90)
```
