## Example

```python
import numpy as np
import pandas as pd

customers = pd.DataFrame({
    'cust_id': [1, 2, 3, 4, 5],
    'name': ['张三','李四','王五','赵六','孙七'],
    'city': ['北京','上海','北京','广州','深圳']})
orders = pd.DataFrame({
    'order_id': [101,102,103,104,105,106],
    'cust_id': [1,1,2,3,3,9],          # 9 是不存在的客户（孤儿订单）
    'product': ['A','B','A','C','B','D'],
    'amount': [100,200,150,300,80,500]})
products = pd.DataFrame({
    'product': ['A','B','C','D'],
    'category': ['食品','电子','服装','家电'],
    'price': [10,200,150,500]})
```

```python
df1 = pd.DataFrame({'x':[1,2]},index=['a','b'])
df2 = pd.DataFrame({'x':[3,4]},index=['c','d'])

pd.concat([df1,df2])
pd.concat([df1,df2],ignore_index=True)

df3 = pd.DataFrame({'y':[3,4]},index=['a','b'])
pd.concat([df1,df3],axis=1)

df4 = pd.DataFrame({'x':[7,8], 'z':[9,10]},index=['e','f'])
pd.concat([df1,df4])
pd.concat([df1,df4],join='inner')
```

```python
orders.merge(customers, on='cust_id', how='inner')
orders.merge(customers, on='cust_id', how='left')
orders.merge(customers, on='cust_id', how='right')
orders.merge(customers, on='cust_id', how='outer')
```

```python
customers['cust_id'].is_unique
orders['cust_id'].is_unique
```

```python
customers2 = customers.rename(columns={'cust_id':'cid'})
orders.merge(customers2, left_on='cust_id', right_on='cid', how = 'inner')

m = orders.merge(customers, on='cust_id', how='outer', indicator= True)
m['_merge'].value_counts()

full = (orders
        .merge(customers, on='cust_id', how='left')
        .merge(products, on='product', how='left'))
full.shape
```

```python
ci = customers.set_index('cust_id')
orders.join(ci,on='cust_id', how='left')
```

```python



customers = pd.DataFrame({
    'cust_id': [1, 2, 3, 4, 5],
    'name': ['张三','李四','王五','赵六','孙七'],
    'city': ['北京','上海','北京','广州','深圳']})
orders = pd.DataFrame({
    'order_id': [101,102,103,104,105,106],
    'cust_id': [1,1,2,3,3,9],          # 9 是不存在的客户（孤儿订单）
    'product': ['A','B','A','C','B','D'],
    'amount': [100,200,150,300,80,500]})
products = pd.DataFrame({
    'product': ['A','B','C','D'],
    'category': ['食品','电子','服装','家电'],
    'price': [10,200,150,500]})

inner_merge = orders.merge(customers, on='cust_id',how='inner')
print(inner_merge.shape[0])
left_merge = orders.merge(customers, on='cust_id',how='left')
print(left_merge.shape[0])
right_merge = orders.merge(customers, on='cust_id',how='right')
print(right_merge.shape[0])
outer_merge = orders.merge(customers, on='cust_id',how='outer')
print(outer_merge.shape[0])

full = orders.merge(customers, on='cust_id',how='left').merge(products, on='product', how='left')
print(full.shape)
```

```python


merge_indicator = orders.merge(customers, on='cust_id',how='outer', indicator=True)
print(merge_indicator[merge_indicator['_merge'] == 'right_only']['name'])
item = merge_indicator[merge_indicator['_merge'] == 'left_only'].merge(products, on='product',how='inner')
print((item['amount']*item['price']).sum())
```
