## Example

```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(1)
stocks = pd.DataFrame({
    'code': ['A','B','C','D','E','F','G','H','I','J'],
    'industry': ['银行','银行','银行','食品饮料','食品饮料','电池','电池','电池','电池','保险'],
    'ret': rng.normal(0, 0.02, 10).round(4),
    'year': [2020,2021,2020,2021,2020,2021,2020,2021,2020,2021]})
```

```python
stocks.groupby('industry')['ret'].agg(mean='mean', max='max', count='count')

stocks.groupby('industry').agg({'ret':'mean','code':'count'})
```

```python
g = stocks.groupby('industry')['ret']
g.agg('mean')
g.transform('mean')
stocks['ret_z'] = (stocks['ret'] - g.transform('mean'))/g.transform('std')
stocks
```

```python
def check(x):
    return len(x)>=3
stocks.groupby('industry').filter(check)
```

```python
stocks.pivot_table(values='ret',index='industry',columns='year',aggfunc='mean',fill_value=None,margins=True)
```

```python
wide = pd.DataFrame({'code':['A','B'], '2020':[0.1,0.2], '2021':[0.3,0.4]})
print(wide)
long = wide.melt(id_vars='code',var_name='year',value_name='ret')
print(long)
long = long.pivot(index='code',columns='year',values='ret')
print(long)
```

```python
stocks = pd.DataFrame({
    'code': ['A','B','C','D','E','F','G','H','I','J'],
    'industry': ['银行','银行','银行','食品饮料','食品饮料','电池','电池','电池','电池','保险'],
    'ret': rng.normal(0, 0.02, 10).round(4),
    'year': [2020,2021,2020,2021,2020,2021,2020,2021,2020,2021]})
g = stocks.groupby('industry')['ret']
stocks['ret_z'] = (stocks['ret'] - g.transform('mean'))/g.transform('std')
stocks.groupby('industry')['ret_z'].agg('mean').round(5) == 0
```

```python
stocks = stocks.groupby('industry').filter(lambda x: len(x)>=3)
g = stocks.groupby('industry')['ret']
stocks['ret_z'] = (stocks['ret'] - g.transform('mean'))/g.transform('std')
stocks
```

```python

stocks.pivot_table(index='industry',columns='year',values='ret',aggfunc='mean',margins=True)
```

```python
stocks = pd.DataFrame({
    'code': ['A','B','C','D','E','F','G','H','I','J'],
    'industry': ['银行','银行','银行','食品饮料','食品饮料','电池','电池','电池','电池','保险'],
    'ret': rng.normal(0, 0.02, 10).round(4),
    'year': [2020,2021,2020,2021,2020,2021,2020,2021,2020,2021]})
wide = stocks.pivot_table(index='code',columns='year',values='ret').reset_index()
long = wide.melt(id_vars='code',var_name='year',value_name='ret').dropna(how='any').reset_index(drop=True)
long
```
