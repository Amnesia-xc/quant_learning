## Example

```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(0)
sales = pd.DataFrame({
    'region': rng.choice(['华东','华北','华南'],12),
    'month':rng.choice([1,2,3],12),
    'sales': rng.integers(100,1000,12)
})
print(sales)
```

```python
g = sales.groupby('region')
print(type(g))
g['sales'].sum()
```

```python
sales.groupby('region')['sales'].sum()
sales.groupby('region')['sales'].mean()

sales.groupby(['region','month'])['sales'].sum().index
# print(type(sales.groupby(['region','month'])['sales'].sum()))
# sales.groupby(['region','month'])['sales'].sum().unstack()
# print(type(sales.groupby(['region','month'])['sales'].sum().unstack()))
```

```python
sales.groupby('region')['sales'].agg(['mean','sum','max','count'])
sales.groupby('region').describe()

sales.groupby('region').size()
sales.groupby('region')['sales'].count()
```

```python
g = sales.groupby('region')
g.get_group('华东')
for name, sub in g:
    print(sub)
    print(name, len(sub))
g.head(2)
```
