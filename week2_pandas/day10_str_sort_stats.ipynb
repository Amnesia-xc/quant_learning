```python
import numpy as np
import pandas as pd
s = pd.Series([' Chicken Bowl ', 'STEAK BURRITO', None, 'Canned Soft Drink'])

s.str.strip()
s.str.upper()
s.str.len()
s.str.startswith(' Chicken')
s.str.contains('Bowl',na=False)
s.str.replace(' ','_',regex=False)
s.str.split(' ',expand=True)
```

```python
chipo = pd.read_csv(
    r'D:\quant_learning\data\chipotle.tsv',
    sep='\t'
)
chipo['choice_description'].str.contains('Rice',na=False).sum()

chipo['choice_description'].str.extract(r'^\[([^,\]]+)')
```

```python
tickets = pd.Series(['订单 A001: 金额 12.3', '订单 A002: 金额 45.6', '订单 A015: 金额 8.8'])
tickets.str.extract(r'A(\d+)')
print(tickets)
tickets.str.extract(r'金额\s*([.\d]+)')
```

```python
u = pd.read_csv(r'D:\quant_learning\data\u.user',sep='|')
u.sort_values('age')
u.sort_values('age',ascending=False)
u.sort_values(['gender','age'],ascending=[True,False])
u.sort_values('age',ignore_index=True)
u.sort_values('age',na_position='first')
u.sort_index()

u.nlargest(3,'age')
u.sort_values('age',ascending=False).head(3)
u.nsmallest(3,'age')
```

```python
u['occupation'].unique()
u['occupation'].nunique()
u['occupation'].value_counts()
u['occupation'].value_counts(normalize=True)
u['occupation'].mode()

chipo['price'] = chipo['item_price'].str.replace('$','',regex=False).str.strip().pipe(pd.to_numeric)
chipo['price'].value_counts(bins=[0,5,10,20,100],sort=False)
```
