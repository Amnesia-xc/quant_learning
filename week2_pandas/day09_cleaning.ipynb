```python
df = pd.DataFrame({
    'code': ['600519','000001','300750','601318'],
    'industry': ['食品饮料', None, '电池', '保险'],
    'pe': [30.0, 5.2, np.nan, 8.1],
})
print(df.isna())
df.isna().sum(axis=0)
df['industry'].fillna('未知')
df['pe'].fillna(df['pe'].mean())
df['pe'].ffill()
df['pe'].bfill()

df.dropna()
df.dropna(how='all')
df.dropna(subset=['industry'])
df.dropna(thresh=2)
```

```python
dirty = pd.DataFrame({
    'code': ['600519','000001','300750','601318','600036','000001','600000'],
    'industry': ['食品饮料','银行','电池',None,'银行','银行','保险'],
    'close': ['1500.0','12.3','210.5','45.6','35.0','12.3','--'],
    'mktcap': [18800,2300,9200,8200,8800,2300,4500],
})
dirty = pd.concat([dirty, pd.DataFrame([dirty.iloc[0].to_dict()])], ignore_index=True)

dirty['industry'] = dirty['industry'].fillna('未知')
dirty['close'] = pd.to_numeric(dirty['close'],errors='coerce')
dirty['close'] = dirty['close'].fillna(dirty['close'].median())
dirty = dirty.drop_duplicates(
    subset=['code'],
    keep='first',
    ignore_index=True
)
dirty.info()
print(dirty.isna().sum())
```
