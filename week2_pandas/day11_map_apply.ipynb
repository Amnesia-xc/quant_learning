## Example

```python
import pandas as pd
import numpy as np

stocks = pd.DataFrame({
    'code': ['600519','000001','300750','601318','600036','000858','002594','601899','600276','000333'],
    'industry': ['食品饮料','银行','电池','保险','银行','食品饮料','汽车','有色金属','医药生物','家用电器'],
    'close': [1500.0,12.3,210.5,45.6,35.0,150.0,250.0,18.0,45.0,70.0],
    'mktcap': [18800,2300,9200,8200,8800,5800,7300,4800,2900,5000]})

stocks['industry_code'] = [2,1,3,4,1,2,6,7,5,8]
code_map = {1:'银行',2:'食品饮料',3:'电池',4:'保险',5:'医药生物',6:'汽车',7:'有色金属',8:'家用电器'}
stocks['industry_name'] = stocks['industry_code'].map(code_map)
stocks['price_level'] = stocks['close'].map(lambda x:'高价' if x > 100 else '低价')
stocks['price_level'] = np.where(stocks['close'] > 100,'高价','低价')

stocks
```

```python
stocks['ratio'] = stocks.apply(lambda row:row['close']/row['mktcap'],axis=1)
stocks['ratio'] = stocks['close']/stocks['mktcap']

stocks[['close','mktcap']].map(lambda x: round(x,1))
stocks[['close','mktcap']].round(0)
```

```python
stocks['new_col'] = 1

stocks = stocks.assign(
    log_mktcap=np.log(stocks['mktcap']),
    scaled=stocks['close']/stocks['close'].max()
)
stocks.insert(0,'date','2026-10-25')
stocks['date'] = pd.to_datetime(stocks['date'])
stocks.info()
```

```python
rng = np.random.default_rng(7)
ret = pd.Series(rng.normal(0,0.02,10))
stocks['is_expensive'] = np.where(stocks['close']>100,1,0)

labels = np.select(
    [ret>0.02,ret>0,ret<-0.02,ret<0],
    ['大涨', '小涨', '大跌', '小跌'],
    default='震荡'
)
ret_df = pd.DataFrame({'ret':ret.round(4),'label':labels})
print(ret_df)
labels
```
