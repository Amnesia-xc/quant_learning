```python
import pandas as pd
import numpy as np
print(pd.__version__)

s = pd.Series([1500.0,12.3,210.5,45.6], name='close')
s = pd.Series([1500.0,12.3,210.5,45.6], 
              index=['600519','000001','300750','601318'],
              name='close')
print(s.values)
print(s.index)
print(s.name)
print(s['601318'])
print(s.iloc[1])
print(s > 100)
print(s[s > 100])
print(s.mean(), s.std())
```
