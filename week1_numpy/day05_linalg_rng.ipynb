## Example

```python
import numpy as np
A = np.array([[1,2],[3,4]])
B = np.array([[5,6],[7,8]])
v = np.array([1,2])

print(A * B)
print(A @ B)
print(A @ v)
print(v @ v)
print(np.dot(v,v))
```

```python
from numpy.linalg import inv, solve, det, eig, qr, cholesky, norm

A = np.array([[2.0,1.0],[1.5,3.0]])
b = np.array([5.0,6.0])

print(inv(A) @ b)
print(solve(A,b))

print(det(A))
print(eig(A))
print(norm(b))

spd = np.array([[4.0,2.0],[2.0,3.0]])
L = cholesky(spd)
print(L@L.T)
```

```python
x = np.array([1.0,2,3,4,5])
y = np.array([2.0,4.1,5.9,8.2,9.7])
print(np.cov(x,y))
print(np.corrcoef(x,y))
```

```python
rng = np.random.default_rng(seed=42)

print(rng.random(5))
print(rng.uniform(-1,1,5))
print(rng.normal(0,1,5))
print(rng.standard_normal(5))
print(rng.exponential(scale=2,size=5))
print(rng.lognormal(0,0.5,5))
print(rng.binomial(n=10,p=0.5,size=5))
print(rng.poisson(lam=3,size=5))
```

```python
rng = np.random.default_rng(42)

print(rng.choice([1,2,3,4,5,6],size=10))
print(rng.choice(["A","B","C"],size=5,p=[0.5,0.3,0.2]))
print(rng.choice(np.arange(10),size=5,replace=False))

x = np.arange(10)
# rng.shuffle(x)
# print(x)
print(rng.permutation(10))
print(x)
```

```python
rng = np.random.default_rng(0)
mean = [0,0]
cov = [[1.0,0.8],[0.8,1.0]]
data = rng.multivariate_normal(mean,cov,size=1000)

print(np.corrcoef(data,rowvar=False))

z = rng.standard_normal((1000000,2))
synthetic = z@np.linalg.cholesky(cov).T
print(np.corrcoef(synthetic,rowvar=False))
```

## Example

```python
rng = np.random.default_rng(0)
A = rng.random((3,3))
M = A.T@A + np.eye(3)
print(inv(M))
print(M@inv(M))
print(np.allclose(np.eye(3),M@inv(M)))
```

```python
b = np.array([1,2,3])
print(solve(M,b))
print(inv(M)@b)
print(np.allclose((solve(M,b)),inv(M)@b))
```

```python
mean = np.array([0.0,0])
cov = np.array([[1,0.8],[0.8,1]])
datac = rng.multivariate_normal(mean,cov,size=1000)
print(np.corrcoef(datac,rowvar=False))
```
