```python
b = rng.random((3,5))
print(b)
print(b.ndim)
print(b.shape)
print(b.size)
print(b.dtype)
v1 = np.zeros(5)
v2 = np.zeros((5,1))
v3 = np.zeros((1,5))
print(v1.shape, v2.shape)
print(v1[0])
print(v2[0])
print(v3[0])
print(v1)
print(v2)
print(v3)
```
